```python
# Activity 1: Basic Stored Procedure – Price Adjustment
# Create a stored procedure update_product_prices that accepts a percentage
# (positive or negative) and updates the price of all products. Add a RAISE NOTICE
# showing how many rows were updated.
# Bonus: Prevent negative prices.

%%sql
CREATE OR REPLACE PROCEDURE update_product_prices(percentage DOUBLE PRECISION)
LANGUAGE plpgsql
AS $$
DECLARE
    rows_updated INTEGER;
BEGIN
    UPDATE products
    SET price = GREATEST(0.01, price * (1 + percentage / 100.0))
    WHERE price > 0;

    GET DIAGNOSTICS rows_updated = ROW_COUNT;
    RAISE NOTICE 'Precos de % produtos atualizados.', rows_updated;
END;
$$;
```

```python
%%sql
CALL update_product_prices(10.1);
```

```python
# Activity 2: Stored Procedure – Category Discount
# Write a procedure apply_category_discount that takes a category name and a
# discount percentage, then applies the discount only to products in that category.
# Log the action in the audit_logs table

%%sql
CREATE OR REPLACE PROCEDURE apply_category_discount(category_name VARCHAR(50), discount_percentage DOUBLE PRECISION)
LANGUAGE plpgsql
AS $$
BEGIN
    UPDATE products
    SET price = price * (discount_percentage / 100.0)
    WHERE category = category_name;

    INSERT INTO audit_logs (action, details)
    VALUES ('discount_applied', CONCAT('Applied ', discount_percentage, '% discount on ', category_name));
END;
$$;
```

```python
%%sql
CALL apply_category_discount('Electronics', 5.0);
```

```python
# Activity 3: Stored Procedure – Order Processing
# Create process_order that receives customer_id and a JSON array of items
# ([{"product_id":1, "quantity":2}, ...]).
# The procedure must:
# ●
# Insert a new order
# ●
# Insert order items
# ● Reduce stock in products
# ● Calculate and update total_amount
# ● Use a single transaction and rollback on error.

%%sql
CREATE OR REPLACE PROCEDURE process_order(p_customer_id INT, p_items JSON)
LANGUAGE plpgsql
AS $$
DECLARE
    v_order_id INT;
    v_total NUMERIC := 0;
    rec JSON;
    v_product_id INT;
    v_quantity INT;
    v_price NUMERIC;
BEGIN
    INSERT INTO orders (customer_id, total_amount)
    VALUES (p_customer_id, 0)
    RETURNING order_id INTO v_order_id;

    FOR rec IN SELECT * FROM json_array_elements(p_items)
    LOOP
        v_product_id := (rec->>'product_id')::INT;
        v_quantity := (rec->>'quantity')::INT;

        SELECT price INTO v_price
        FROM products
        WHERE product_id = v_product_id;

        IF v_price IS NULL THEN
            RAISE EXCEPTION 'Produto % nao encontrado', v_product_id;
        END IF;

        INSERT INTO order_items (order_id, product_id, quantity, unit_price)
        VALUES (v_order_id, v_product_id, v_quantity, v_price);

        UPDATE products
        SET stock = stock - v_quantity
        WHERE product_id = v_product_id;

        v_total := v_total + (v_price * v_quantity);
    END LOOP;

    UPDATE orders
    SET total_amount = v_total
    WHERE order_id = v_order_id;

EXCEPTION
    WHEN OTHERS THEN
        RAISE;
END;
$$;
```

```python
# Activity 4: Stored Procedure – Low Stock Alert
# Build notify_low_stock that finds products with stock < 10, inserts a row into
# audit_logs, and raises a notice with the list of products.

%%sql
CREATE OR REPLACE PROCEDURE notify_low_stock()
LANGUAGE plpgsql
AS $$
DECLARE
    rec RECORD;
    product_count INT := 0;
    log_message TEXT := '';
BEGIN
    FOR rec IN
        SELECT product_id, stock
        FROM products
        WHERE stock < 10
    LOOP
        product_count := product_count + 1;
        log_message := CONCAT('Product ID ', rec.product_id, ' stock below 10 ', rec.stock);
        INSERT INTO audit_logs (action, details)
        VALUES ('low_stock', log_message);
        RAISE NOTICE '%', log_message;
    END LOOP;
    IF product_count = 0 THEN
        RAISE NOTICE 'Nenhum produto com baixo estoque encontrado.';
    END IF;

END;
$$;
```

```python
# Activity 5: Scalar Function – Final Price
# Create a function calculate_final_price(product_id INTEGER, quantity INTEGER)
# that returns the price after applying:
# ● 10 % tax
# ● 5 % loyalty discount if the customer has > 500 points (you may add a parameter
# for customer_id).
# Return type: NUMERIC.

%%sql
CREATE OR REPLACE FUNCTION calculate_final_price(product_id_param INTEGER, quantity_param INTEGER, customer_id_param INTEGER)
RETURNS NUMERIC
AS $$
DECLARE
  unit_price NUMERIC;
  loyalty_points_var INTEGER;
  price_after_tax NUMERIC;
  final_unit_price NUMERIC;
BEGIN
  SELECT price INTO unit_price FROM products WHERE product_id = product_id_param;
  price_after_tax = unit_price * 1.10;
  SELECT loyalty_points INTO loyalty_points_var FROM customers WHERE customer_id = customer_id_param;
  IF loyalty_points_var > 500 THEN
    final_unit_price = price_after_tax * 0.95;
  ELSE
    final_unit_price = price_after_tax;
  END IF;
  RETURN final_unit_price * quantity_param;
END;
$$ LANGUAGE plpgsql;
```

```python
# Activity 6: Set-Returning Function – Top Sellers
# Develop a table function top_selling_products(days INTEGER) that returns a table
# with the top 10 best-selling products (by total quantity sold) in the last N days.

%%sql
CREATE OR REPLACE FUNCTION top_selling_products(days INTEGER)
RETURNS TABLE (
    product_id INTEGER,
    product_name VARCHAR,
    total_quantity_sold BIGINT
)
LANGUAGE plpgsql
AS $$
BEGIN
    RETURN QUERY
    SELECT
        p.product_id,
        p.name AS product_name,
        SUM(oi.quantity) AS total_quantity_sold
    FROM
        products p
    JOIN
        order_items oi ON p.product_id = oi.product_id
    JOIN
        orders o ON oi.order_id = o.order_id
    WHERE
        o.order_date >= NOW() - INTERVAL '1 day' * days
    GROUP BY
        p.product_id, p.name
    ORDER BY
        total_quantity_sold DESC
    LIMIT 10;
END;
$$;
```

```python
# Activity 7: Function – Customer Lifetime Value
# Write a function customer_ltv(customer_id INTEGER) that returns the total
# amount spent by that customer across all completed orders.

%%sql
CREATE OR REPLACE FUNCTION customer_ltv(customer_id_param INTEGER)
RETURNS NUMERIC
AS $$
DECLARE
    ltv_amount NUMERIC;
BEGIN
    SELECT SUM(total_amount)
    INTO ltv_amount
    FROM orders
    WHERE customer_id = customer_id_param AND status = 'completed';

    RETURN COALESCE(ltv_amount, 0);
END;
$$ LANGUAGE plpgsql;
```

```python
# Activity 8: Scheduled Job – Weekly Expiry Discount
# Create a procedure apply_expiry_discount that gives a 20 % discount to any
# product expiring in the next 7 days.
# Then schedule it with pg_cron to run every Monday at 2:00 AM.

%%sql
CREATE OR REPLACE PROCEDURE apply_expiry_discount()
LANGUAGE plpgsql
AS $$
DECLARE
    rows_updated INTEGER;
BEGIN
    UPDATE products
    SET price = GREATEST(0.01, price * 0.80)
    WHERE expiry_date IS NOT NULL
      AND expiry_date <= NOW() + INTERVAL '7 days'
      AND price > 0;

    GET DIAGNOSTICS rows_updated = ROW_COUNT;
END;
$$;
```

```python
%%sql
SELECT cron.schedule(
    'weekly_expiry_discount',
    '0 2 * * 1',
    $$
    CALL apply_expiry_discount();
    $$
);
```

```python
# Activity 9: Scheduled Job – Monthly Log Cleanup
# Create a procedure cleanup_old_audit_logs that deletes records from audit_logs
# older than 1 year.
# Schedule it with pg_cron to run on the 1st day of every month at 3:00 AM.

%%sql
CREATE OR REPLACE PROCEDURE cleanup_old_audit_logs()
LANGUAGE plpgsql
AS $$
DECLARE
    rows_deleted INTEGER;
BEGIN
    DELETE FROM audit_logs
    WHERE log_date < NOW() - INTERVAL '1 year';
    GET DIAGNOSTICS rows_deleted = ROW_COUNT;
END;
$$;
```

```python
%%sql
SELECT cron.schedule(
    'monthly_audit_cleanup',
    '0 3 1 * *',
    $$
    CALL cleanup_old_audit_logs();
    $$
);
```

```python
# Activity 10: Integrated Capstone Project
# Combine everything:
# Create a procedure daily_ecommerce_report that:
# ● Calculates total sales of the previous day
# ● Updates loyalty points for every customer who made a purchase
# ● Logs everything in audit_logs
# Schedule this procedure with pg_cron to run every day at 6:00 AM.
# Add error handling and a final RAISE NOTICE with the summary.

%%sql
CREATE OR REPLACE PROCEDURE daily_ecommerce_report()
LANGUAGE plpgsql
AS $$
DECLARE
    total_sales NUMERIC := 0;
    rows_updated INTEGER;
    report_date DATE := CURRENT_DATE - INTERVAL '1 day';
BEGIN
    SELECT COALESCE(SUM(total_amount), 0)
    INTO total_sales
    FROM orders
    WHERE DATE(order_date) = report_date;

    UPDATE customers c
    SET loyalty_points = loyalty_points + FLOOR(o.total_amount / 10)
    FROM orders o
    WHERE c.customer_id = o.customer_id
      AND DATE(o.order_date) = report_date;

    GET DIAGNOSTICS rows_updated = ROW_COUNT;

    INSERT INTO audit_logs (log_date, action, details)
    VALUES (
        NOW(),
        'DAILY_REPORT',
        'Total sales: ' || total_sales ||
        ', Customers updated: ' || rows_updated
    );
END;
$$;
```

```python
%%sql
SELECT cron.schedule(
    'daily_ecommerce_report_job',
    '0 6 * * *',
    $$
    CALL daily_ecommerce_report();
    $$
);
```
